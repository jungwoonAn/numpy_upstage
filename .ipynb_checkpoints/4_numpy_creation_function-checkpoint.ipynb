```python
import numpy as np
```

### arange
- array의 범위를 지정하여, 값의 list를 생성하는 명령어
- arange(start,end,step)

```python
list(range(0,10,3))
```

```python
type(list(range(30)))
```

```python
# range: List의 range와 같은 효과,
# integer로 0부터 20까지 배열 추출
np.arange(30)
```

```python
np.arange(0,10,0.5)
```

```python
# floating point도 표시가능함
np.arange(0,5,0.5)
```

```python
np.arange(30).reshape(5,6)
```

### ones, zeros & empty
- zeros: 0으로 가득찬 ndarray 생성
    - np.zeros(shape,dtype,order)
- ones: 1로 가득찬 ndarray 생성
    - np.ones(shape, dtype,order)
- empty: shape만 주어지고 비어있는 ndarray 생성 (memory initialization이 되지 않음)
- something_like: 기존 ndarray의 shape 크기 만큼 1,0 또는 emptyarray를 반환

```python
# zero vector 생성
np.zeros(shape=(10,), dtype=np.int8)
```

```python
# 2 by 5 - zero matrix 생성
np.zeros((2,5))
```

```python
np.ones(shape=(10,), dtype=np.int8)
```

```python
np.ones((2,5))
```

```python
np.empty(shape=(10,), dtype=np.int8)
```

```python
np.empty((3,5))
```

```python
test_matrix = np.arange(30).reshape(5,6)
np.ones_like(test_matrix)
```

```python
np.zeros_like(test_matrix)
```

### identity
- 단위 행렬(identity matrix)를 생성함

```python
np.identity(n=3, dtype=np.int8)
```

```python
np.identity(n=5)
```

### eye
- 대각선이 1인 행렬, k값의 시작 index의 변경이 가능

```python
np.eye(3)
```

```python
np.eye(N=3, M=5, dtype=np.int8)
```

```python
np.eye(3,5,k=2)
```

### diag
- 대각 행렬의 값을 추출함

```python
matrix = np.arange(9).reshape(3,3)
np.diag(matrix)
```

```python
np.diag(matrix, k=1)  # k는 start index
```

### numpy.random
- 데이터 분포에 따른 sampling으로 array를 생성

```python
# 균등 분포
np.random.uniform(0,1,10).reshape(2,5)
```

```python
# 정규 분포
np.random.normal(0,1,10).reshape(2,5)
```
