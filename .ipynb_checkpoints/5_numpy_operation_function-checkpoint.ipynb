```python
import numpy as np
```

### operation in array
- sum: ndarray의 element들 간의 합을 구함, list의 sum기능과 동일

```python
test_array = np.arange(1,11)
test_array
```

```python
test_array.sum()
```

```python
test_array.sum(dtype=np.float32)
```

- axis: 모든 operation function을 실행할 때 기준이 되는 dimension 축

```python
test_array = np.arange(1,13).reshape(3,4)
test_array
```

```python
test_array.sum(axis=1), test_array.sum(axis=0)
```

```python
third_order_tensor = np.array([test_array, test_array, test_array])
third_order_tensor
```

```python
third_order_tensor.sum(axis=2)
```

```python
third_order_tensor.sum(axis=1)
```

```python
third_order_tensor.sum(axis=0)
```

- mean & std: ndarray의 element들 간의 평균 또는 표준편차를 반환

```python
test_array
```

```python
test_array.mean(), test_array.mean(axis=0)
```

```python
test_array.std(), test_array.std(axis=0)
```

- mathematical functions: 그 외에도 다양한 수학 연산자를 제공함
    - exponential: exp, expml, exp2, log, log10, loglp, log2, power, sqrt
    - trigonometric: sin, cos, tan, acsin, arccos, atctan
    - hyperbolic: shnh, cosh, tanh, acsinh, arccosh, atctanh

```python
np.exp(test_array), np.sqrt(test_array)
```

### concatenate
- NumPy array를 합치는(붙이는) 함수

```python
a = np.array([1,2,3])
b = np.array([2,3,4])
np.vstack((a,b))
```

```python
a = np.array([[1],[2],[3]])
b = np.array([[2],[3],[4]])
np.hstack((a,b))
```

```python
a = np.array([[1,2,3]])
b = np.array([[2,3,4]])
np.concatenate((a,b), axis=0)
```

```python
a = np.array([[1,2], [3,4]])
b = np.array([[5],[6]])
np.concatenate((a,b), axis=1)
```

```python
a = np.array([[1,2], [3,4]])
b = np.array([[5,6]])
np.concatenate((a,b.T), axis=1)
```

```python
a = np.array([[1,2], [3,4]])
b = np.array([5,6])
print(b)
b = b[np.newaxis, :]
print(b)
np.concatenate((a,b.T), axis=1)
```
