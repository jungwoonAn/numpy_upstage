```python
import numpy as np
```

### reshape
- reshape: Array의 shape을 변경함, element의 개수는 동일
- -1: size를 기반으로 row 개수 선정

```python
test_matrix = [[1,2,3,4],[1,2,5,8]]
np.array(test_matrix).shape
```

```python
np.array(test_matrix).reshape(4,2)
```

```python
np.array(test_matrix).reshape(2,2,2)
```

```python
# 원래 데이터인 test_matrix는 변경되지 않음
test_matrix
```

```python
np.array(test_matrix).reshape(8)
```

```python
np.array(test_matrix).reshape(2,4).shape
```

```python
np.array(test_matrix).reshape(2,-1)
```

```python
np.array(test_matrix).reshape(1,-1,2)
```

### flat or flatten()
- flatten : 다차원 array를 1차원 array로 변환

```python
test_matrix = [[[1,2,3,4],[1,2,5,8]],[[1,2,3,4],[1,2,5,8]]]
np.array(test_matrix)
```

```python
test_matrix = np.array(test_matrix)
test_matrix.shape
```

```python
test_matrix.flatten()
```

```python
test_matrix.flatten().size
```

```python
test_matrix.flatten().shape
```
