### numpy dtype
- ndarray의 single element가 가지는 data type
- 각 element가 차자히는 memory의 크기가 결정됨

```python
a = np.array([[1,2,3],[4.5,5,6]], dtype=np.int8)
```

```python
a.dtype
```

- nbytes: ndarray object의 메모리 크기를 반환함
- 8bits = 1bytes
- 32bits = 4bytes
- 64bits = 8bytes

```python
np.array([[1,2,3],[4.5,'5','6']], dtype=np.float32).nbytes  # 6*4bytes
```

```python
np.array([[1,2,3],[4.5,'5','6']], dtype=np.int8).nbytes  # 6*1bytes
```

```python
np.array([[1,2,3],[4.5,'5','6']], dtype=np.float64).nbytes  # 6*8bytes
```
